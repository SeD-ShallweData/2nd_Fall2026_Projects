**입력:** 완료된 `flow_age_pop` 결합 캐시의 남녀·제공 전체 연령 합계. 강남구·춘천시를 분리하고 2025년 7~12월 6개월 모두 유효한 기존 `COMMON_VALID` 위치만 연결합니다. 보류·중복·지역 배정 변경 정책과 정규화 위치 키를 유지합니다. 시간대·요일 자료를 합산하지 않습니다.

**계산:** 전월/7월 대비 변화량·변화율, 현재/최장 연속 감소 구간 수, 현재/최장 7월 아래 연속 유지 월수. 같은 값이면 연속 감소가 끊기며, 7월 이상으로 회복하면 기준월 아래 유지가 끊깁니다. 7월 전월 비교는 미산출, 분모 0의 변화율은 미산출입니다. 0·작은 값·극단적인 변화율은 제거하거나 보정하지 않습니다.

**행정동 요약:** 고정 위치의 유동 합계 변화율, 전월/7월 감소 위치 비율, 현재 지속기간 0~5의 정확한 위치 수·비율 분포, 감소 위치가 7월 유동에서 차지한 비중. 위치 비율의 분모는 고정 비교 위치 수이며 7월 전월 비교 가능 수는 0입니다. 유동 비중의 분모는 같은 위치 집합의 7월 유동 합계입니다. 무관측과 합계 0을 구분합니다. 행정동 변화율은 위치별 변화율의 평균이 아닙니다.

## 실험 ② · 연령대별 월간 유동 변화

**질문:** 같은 위치 집합에서 어느 연령대의 유동 수준과 구성비가 더 크게 변화하는가? 실험 ①의 전체 감소와 연령별 변화는 어떻게 다른가?

**입력:** `flow_age_pop`의 남녀 합산 10대·20대·30대·40대·50대·60대 이상. 강남구·춘천시를 분리하고 2025년 7~12월 기존 `COMMON_VALID` 고정 위치 집합을 모든 연령에 동일하게 적용합니다. 원본·매핑·실험 ①·주민등록인구를 갱신하지 않습니다.

**계산:** 실험 ①의 `position_monthly_changes`를 각 연령에 적용해 위치별 월간 연결·변화·지속성을 계산한 뒤 행정동과 시군구로 요약합니다. 지역 합계에도 같은 함수로 전월/7월 변화와 현재·최장 연속 감소 구간 수 및 7월 아래 유지 월수를 계산합니다. 위치별 현재 지속기간 분포는 별도 열로 제공하며 지역 합계의 지속성과 구분합니다. 연령별 위치 중복 CSV는 저장하지 않습니다.

**구성비:** 연령 유동 합계 / 동일 위치 전체 연령 합계 × 100. 구성비의 전월/7월 변화는 변화율이 아닌 `%p`입니다. 0·작은 값은 보존하고, 연령 변화율의 0 분모 및 구성비의 전체 합계 0은 각각 미산출 상태로 표시합니다. 7월 전월 비교는 미산출입니다.

**비교:** 6개 연령의 원래 수준·변화율·구성비를 나란히 표시하고, 같은 연령의 행정동 변화율과 해당 시군구 전체 변화율 차이를 `%p`로 제공합니다. 시군구 값에는 선택 행정동이 포함되므로 독립 대조군이 아닙니다. 전체 변화율에 대한 각 연령의 기여도(연령 변화량 / 전체 비교월 합계 × 100)는 구성비 변화와 구분합니다. 6개 연령의 기여도 합은 전체 변화율입니다.

**실행·결과:** 새 커널에서 이 절의 셀 3개만 독립 실행하거나 이미 읽은 위치 메모리를 재사용합니다. 실험 ① 계산·내보내기는 실행하지 않습니다. 요약 CSV는 `outputs/canonical/exports/실험2_연령대별_월간유동변화.csv`에 저장합니다. 마지막 셀에서 지역·월·행정동·지도 연령을 선택합니다. 기본값은 제공 연령 순서·행정동 이름 순서의 첫 항목이며 타겟 선정이 아닙니다. 그래프와 표는 모든 연령을 비교합니다.

연령별 차이는 관측 집계자료의 기술통계입니다. 통계적 유의성·개인의 고립·최종 연령 타겟을 판정하지 않습니다. 주민등록인구와 외부 수집이 필요하지 않습니다.

**질문:** 평일·주말 중 어느 쪽의 수준 변화가 더 큰가? 주말 감소와 평일 증가로 인한 주말/평일 비율 감소는 어떻게 다른가?

**정의 확인:** 제공 Excel 정의서 `Sheet1!C72`는 ‘요일단위 요일별 일평균 유동인구’이며 `B78:B84`와 CSV의 실제 원본 열 `FLOW_POP_CNT_MON/TUS/WED/THU/FRI/SAT/SUN`이 일치합니다. 입력 읽기 전에 같은 정의를 확인하며, 확인하지 못하면 평균식을 임의 적용하지 않습니다.
